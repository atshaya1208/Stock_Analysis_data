```python
import pandas as pd
import numpy as np
import os
import yaml
```

```python
search_path = "/content/stock_data"

yaml_files = []

for root, dirs, files in os.walk(search_path):
    for file in files:
        if file.lower().endswith(('.yaml')):
            full_path = os.path.join(root, file)
            yaml_files.append(full_path)

print(f"Found {len(yaml_files)} YAML files:")
```

```python
for file in yaml_files:
    print(file)
```

```python

all_dataframes = []

for filepath in yaml_files:
    with open(filepath, 'r') as file:
      content = yaml.safe_load(file)
      df = pd.DataFrame([content])

      df['source_file'] = os.path.basename(filepath)
      all_dataframes.append(df)

if all_dataframes:
    data = pd.concat(all_dataframes, ignore_index=True)
    print("DataFrame Shape:", data.shape)
else:
    print("No data found")

data.head()
```

```python
split_dfs = {}

normalized_data = pd.json_normalize(all_stock_data)

for symbol, group in normalized_data.groupby('Ticker'):
    split_dfs[symbol] = group
```

```python
df_list = ['SBIN', 'BAJFINANCE', 'TITAN', 'ITC', 'TCS', 'LT', 'TATACONSUM', 'RELIANCE', 'HCLTECH', 'JSWSTEEL', 'ULTRACEMCO', 'POWERGRID', 'INFY', 'TRENT', 'BHARTIARTL', 'TATAMOTORS', 'WIPRO', 'TECHM', 'NTPC', 'HINDUNILVR', 'APOLLOHOSP', 'M&M', 'GRASIM', 'ICICIBANK', 'ADANIENT', 'ADANIPORTS', 'BEL', 'BAJAJFINSV', 'EICHERMOT', 'COALINDIA', 'MARUTI', 'INDUSINDBK', 'ASIANPAINT', 'TATASTEEL', 'HDFCLIFE', 'DRREDDY', 'SUNPHARMA', 'KOTAKBANK', 'SHRIRAMFIN', 'NESTLEIND', 'ONGC', 'CIPLA', 'BPCL', 'BRITANNIA', 'SBILIFE', 'HINDALCO', 'HEROMOTOCO', 'AXISBANK', 'HDFCBANK', 'BAJAJ-AUTO']

def clean_dfs(df):

  df = df.drop_duplicates()

  df.columns = [c.lower().replace(' ', '_') for c in df.columns]

  df.ticker = df.ticker.str.lower().str.strip()

  if 'date' in df.columns:
      df['date'] = pd.to_datetime(df['date'])
      df.set_index('date', inplace=True)
      df = df.sort_index()

  return df

cleaned_dfs = {name: clean_dfs(df) for name, df in split_dfs.items()}
```

```python
# volatility analysis

volatility_list = []

for name,df in cleaned_dfs.items():

  df['prev_close'] = df['close'].shift()
  df['daily_returns'] = (df['close'] - df['prev_close'])/ df['prev_close']

  clean_returns = df['daily_returns'].dropna()

  stock_volatility = np.std(clean_returns, ddof=1)

  volatility_list.append((name,stock_volatility))

volatility_df = pd.DataFrame(volatility_list)
```

```python
volatility_df
```

```python
volatility_df = volatility_df.rename(columns ={0:'ticker', 1:'volatility'})
sorted_vol_df = volatility_df.sort_values(by='volatility', ascending = False).head(10)
sorted_vol_df
```

```python
Volatility = sorted_vol_df.to_csv('Volatility.csv', index = False)
```

```python
# cumulative return over time

cumulative_returns = []

for name,df in cleaned_dfs.items():

  first_open = df['open'].iloc[0]
  last_close = df['close'].iloc[-1]

  per_diff = ((last_close - first_open)/first_open)*100

  cumulative_returns.append((name,per_diff))

cumulative_df = pd.DataFrame(cumulative_returns)
```

```python
cumulative_df
```

```python
cumulative_df = cumulative_df.rename(columns ={0:'ticker', 1:'cum_percentage'})
sorted_cum_df = cumulative_df.sort_values(by='cum_percentage', ascending = False).head(5)
sorted_cum_df
```

```python
Cumulative_percentage = sorted_cum_df.to_csv('Cumulative_percentage.csv', index = False)
```

```python
#month wise gainers and losers

all_stocks_df = pd.concat(cleaned_dfs.values(), ignore_index=True)

monthly_stock = all_stocks_df.groupby(['ticker', 'month']).agg(open=('open', 'first'), close=('close', 'last'))
monthly_stock['monthly_returns'] = (monthly_stock['close'] - monthly_stock['open']) / monthly_stock['open']
monthly_returns_df = monthly_stock.reset_index()
```

```python
monthly_returns_df
```

```python
monthly_returns_df.info()
```

```python
split_monthdata = {}

normalizing_data = monthly_returns_df

for ticker, group in normalizing_data.groupby('month'):
    split_monthdata[ticker] = group
```

```python
split_monthdata['2023-10']
```

```python
result_data = {}

for month,df in split_monthdata.items():
  gainers = df.nlargest(5, 'monthly_returns')
  losers = df.nsmallest(5, 'monthly_returns')

  monthly_gl_df = pd.concat([gainers, losers]).sort_values('monthly_returns', ascending=False)


  result_data[month] = monthly_gl_df

```
